# Variance reduction

Standard error of a European call under four sampling schemes on the same seed
and path count: IID, antithetic, control variate, and both.

```python
from monte_carlo_option_engine import (
    Contract,
    ContractKind,
    Market,
    black_scholes,
    price_mc,
)

market = Market(S=100, T=0.5, r=0.04, q=0.01, sigma=0.25)
call = Contract(K=105, kind=ContractKind.euro_call)
bs = black_scholes(market, call)
n = 8_000
schemes = [
    ("IID", False, False),
    ("antithetic", True, False),
    ("control variate", False, True),
    ("antithetic + CV", True, True),
]

print(f"BS {bs:.6f}   n={n}\n")
for label, anti, cv in schemes:
    result = price_mc(
        market,
        call,
        trial_count=n,
        seed=0,
        antithetic=anti,
        control_variate=cv,
    )
    print(
        f"{label:<18} price={result.price:.6f}  stderr={result.stderr:.6f}  "
        f"|error|={abs(result.price - bs):.6f}"
    )
```
